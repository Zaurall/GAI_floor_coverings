```python
import os
import cv2
import shutil
from image_gen_aux import UpscaleWithModel
```

```python
input_folder = '../data/preprocessed/resized'
output_folder = '../data/preprocessed/upscaled'

# Remove all previous files in directory
if os.path.exists(output_folder):
    shutil.rmtree(output_folder)
# Make an output directory
os.makedirs(output_folder, exist_ok=True)
```

```python
# Initialize the Real-ESRGAN upscaler
upscaler = UpscaleWithModel.from_pretrained("Kim2091/UltraSharp").to("cuda")
```

```python
# Process images in each input folder
for filename in sorted(os.listdir(input_folder)):
    if filename.endswith('.png') or filename.endswith('.jpg') or filename.endswith('.webp'):
        file_path = os.path.join(input_folder, filename)
        output_file_path = os.path.join(output_folder, filename)
        
        # Read the image
        image = cv2.imread(file_path, cv2.IMREAD_UNCHANGED)
        if image is None:
            print(f"Warning: Could not read image {file_path}. Skipping.")
            continue
        
        # Upscale and resize the image
        processed_image = upscaler(image)
        
        # Save the processed image
        cv2.imwrite(output_file_path, processed_image)

print("All images have been processed and saved.")
```
