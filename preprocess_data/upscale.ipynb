```python
import os
import cv2
import torch
import shutil
import zipfile
from tqdm import tqdm
from PIL import Image
from diffusers import StableDiffusionUpscalePipeline
```

```python
input_folder = '../data/preprocessed/resized'
output_folder = '../data/preprocessed/upscaled'

# Remove all previous files in directory
# if os.path.exists(output_folder):
#     shutil.rmtree(output_folder)
# Make an output directory
os.makedirs(output_folder, exist_ok=True)
```

```python
# load model and scheduler
model_id = "stabilityai/stable-diffusion-x4-upscaler"
pipeline = StableDiffusionUpscalePipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipeline = pipeline.to("cuda")
```

```python
processed_images = os.listdir(output_folder)

# Process images in each input folder
for filename in tqdm(sorted(os.listdir(input_folder))):
    if filename.endswith('.png') or filename.endswith('.jpg') or filename.endswith('.webp'):
        if filename not in processed_images:
            file_path = os.path.join(input_folder, filename)
            output_file_path = os.path.join(output_folder, filename)
            
            # Read the image
            try:
                image = Image.open(file_path).convert("RGB")  # Ensure image is in RGB format
            except Exception as e:
                print(f"Warning: Could not read image {file_path}. Skipping. Error: {e}")
                continue
    
            width, height = image.size
            new_height = 512
            new_width = int(width * (new_height / height))
            resized_image = image.resize((new_width, new_height), Image.LANCZOS)
    
            
            processed_image = pipeline(prompt="", image=resized_image).images[0]
            processed_image.save(output_file_path)

print("All images have been processed and saved.")
```

```python
# Create a zip file of all images in the output folder
zip_file_path = '../data/preprocessed/upscaled.zip'
with zipfile.ZipFile(zip_file_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, files in os.walk(output_folder):
        for file in files:
            file_path = os.path.join(root, file)
            arcname = os.path.relpath(file_path, output_folder)
            zipf.write(file_path, arcname)
```
